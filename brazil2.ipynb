 Consumer Preference Analysis



By comparing isBestSeller, reviews, and stars, you can analyze the differences between best-selling products and non-best-sellers.
For example, you can check if products with isBestSeller == TRUE have significantly more reviews or higher ratings compared to those with FALSE.
Analysis Questions:

Do best-selling products generally have more reviews?

Do best-selling products have higher ratings compared to non-best-sellers?


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind

df = pd.read_csv('/content/amz_br_datacleaned.csv')

df.head(5)
```

```python
# Convert 'isBestSeller' to numerical (0 = False, 1 = True)
df['isBestSeller_num'] = df['isBestSeller'].astype(int)

# Select relevant columns for correlation matrix
correlation_columns = ['reviews', 'stars', 'price', 'isBestSeller_num']
correlation_matrix = df[correlation_columns].corr()

# Display the correlation matrix
print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix, annot=True, cmap='Purples', linewidths=0.5, fmt='.2f')

plt.title('Correlation Heatmap: Reviews, Ratings, Price, and Best Seller Status')
plt.show()
```

Reviews and isBestSeller_num (0.02): There is very weak positive correlation between the number of reviews and whether the product is a best-seller. This suggests that the number of reviews doesn’t significantly impact whether a product becomes a best-seller.

  
  2.	Stars and isBestSeller_num (0.06): Similarly, there’s a very weak positive correlation between star ratings and best-seller status. Best-sellers may have slightly higher ratings, but the relationship is quite small.

  
  3.	Price and isBestSeller_num (-0.01): There’s a very weak negative correlation between price and best-seller status, meaning price has no real influence on whether a product is a best-seller.

  
  4.	Stars and Reviews (0.06): Star ratings and the number of reviews have a very weak positive correlation, suggesting little to no linear relationship between them.


The weak correlations suggest that factors like the number of reviews, star ratings, and price do not strongly influence a product’s best-seller status. You may want to explore other variables (e.g., product category, brand, or specific features) to see if they have a stronger influence on being a best-seller.
