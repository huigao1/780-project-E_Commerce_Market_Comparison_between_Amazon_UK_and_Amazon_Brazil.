```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind

# Load the dataset
df = pd.read_csv('/content/amz_uk_datacleaned.csv')

# Preview the dataset
df.head(5)
```

```python
df['isBestSeller_num'] = df['isBestSeller'].astype(int)

correlation_columns = ['reviews', 'stars', 'price', 'isBestSeller_num']
correlation_matrix = df[correlation_columns].corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix, annot=True, cmap='Purples', linewidths=0.5, fmt='.2f')

plt.title('Correlation Heatmap: Reviews, Ratings, Price, and Best Seller Status')
plt.show()
```

Reviews and Best Seller Status (0.09):
	•	There is a weak positive correlation between the number of reviews and the likelihood of being a best-seller. This suggests that products with more reviews have a slightly better chance of being best-sellers, though the relationship is not strong.

  
  
  2.	Stars and Best Seller Status (0.06):
	•	The correlation between star ratings and best-seller status is also weak positive. Products with higher ratings are slightly more likely to be best-sellers, but the influence is small.

  
  3.	Price and Best Seller Status (-0.01):
	•	There is an extremely weak negative correlation between price and being a best-seller. Price has almost no influence on whether a product becomes a best-seller in this dataset.

  
  4.	Stars and Reviews (0.12):
	•	A weak positive correlation exists between star ratings and the number of reviews. Products with more reviews tend to have slightly higher ratings.

  
  5.	Price and Stars (-0.13):
	•	There is a weak negative correlation between price and star ratings, meaning higher-priced products may have slightly lower ratings, but this effect is not strong.
